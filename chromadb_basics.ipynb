```python
import chromadb
```

```python
client=chromadb.Client()
```

```python
collection=client.create_collection(name="news")
```

```python
collection.add(
    ids=["id1", "id2", "id3"],
    documents=["lorem ipsum...", "doc2", "doc3"],
    metadatas=[{"chapter": 3, "verse": 16}, {"chapter": 3, "verse": 5}, {"chapter": 29, "verse": 11}],
)
```

```python
results=collection.query(
    query_texts="lorem",
    n_results=1
)
results
```

```python
objects=collection.peek()
```

```python
len(objects)
```

```python
from datetime import datetime
collection1=client.create_collection(
    name='reviews',
    metadata={
        "description":"Product reviews",
        "created":str(datetime.now())

        
    }
)
```

```python
collection1.add(
    documents=[
    "Absolutely love this product! Exceeded all my expectations and works great.",
    "Decent quality for the price, but shipping took much longer than expected.",
    "Stopped working after just two weeks of light use. Very disappointed.",
    "Customer service was fantastic when I had a question, and the item is top-notch."
    ],
    ids=["r1","r2","r3","r4"],
    metadatas=[
    {"category": "electronics", "rating": 5},
    {"category": "electronics", "rating": 3},
    {"category": "electronics", "rating": 1},
    {"category": "customer_service", "rating": 5}
    ]
)
```

```python
collection1.count()
```

```python
collection1.peek()
```

```python
collection1.query(
    query_texts=["product is bad"],
    n_results=2
)
```

```python
collection1.query(
    query_texts=["product is bad"],
    n_results=2,
    where={"category":'electronics'}
)
```

```python
collection1.delete(ids=['r1'])
```

```python
collection1.query(
     query_texts=["slow delivery"],
    n_results=2,
    where={"rating":{"$in":[1,2,3]}} 
)
```

```python
collection1.query(
     query_texts=["slow delivery"],
    n_results=2,
    where={"rating":{"$lt":3}} 
)
```
