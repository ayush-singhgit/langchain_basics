```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.exceptions import OutputParserException
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
prompt='''
Extract revenue and eps in json format in 4 diffexerent keys from the givn below news article.

Article
=======
{article}
'''

pt=PromptTemplate.from_template(prompt)
pt
```

```python
llm=ChatGroq(model_name="openai/gpt-oss-120b")
chain= pt | llm
```

```python
article_text='''
echGiant Inc. reported a strong third quarter with total revenue surging to $4.5 billion, beating analyst expectations.
Net income saw a steady increase, driven by robust cloud services growth and strong enterprise software adoption.
Consequently, the diluted earnings per share (EPS) climbed to $1.25, up significantly from $1.10 in the same quarter last year.
Management remains optimistic about the upcoming fiscal year, projecting continued margin expansion and steady global demand.

'''
res=chain.invoke({"article":article_text})
res.content
```

```python
parsor=JsonOutputParser()
parsor.parse(res.content)
```
