```python
from langchain_ollama import ChatOllama

llm=ChatOllama(model="llama3.2:3b", temperature=1.8,num_predict=100)
response=llm.invoke("hello, what's up!")
print(response.content)
```

### Prompt template

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template("Write a two line poem on {topic}")

prompt.invoke({"topic":"cats"})
```

```python
llm=ChatOllama(model="llama3.2:3b", temperature=1.8,num_predict=100)
```

```python
chain=prompt | llm
response=chain.invoke({"topic": "dogs"})
print(response.content)
```
